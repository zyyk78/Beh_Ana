```python
fpath='/home/zyyk78/HDD/zyy_rec/Stage2/A02_PPC/20241217_01.pkl'
import pickle
with open(fpath, 'rb') as file:
    behave = pickle.load(file)
```

```python
#score
scoreL=len(behave['SlnL']['Time_S'])/(len(behave['ROmL']['Time_S'])+len(behave['SlnL']['Time_S']))
scoreR=len(behave['SlnR']['Time_S'])/(len(behave['ROmR']['Time_S'])+len(behave['SlnR']['Time_S']))
score=(len(behave['SlnR']['Time_S'])+len(behave['SlnL']['Time_S']))/(len(behave['ROmL']['Time_S'])+len(behave['SlnL']['Time_S'])+len(behave['ROmR']['Time_S'])+len(behave['SlnR']['Time_S']))
print("L ",str(scoreL)," R ",str(scoreR)," Tot ",score)
```

```python
#事件排序
eve_list=[]
for timestamp in behave['SlnL']['Time_S']:
    eve_list.append((timestamp,'RWR'))
for timestamp in behave['SlnR']['Time_S']:
    eve_list.append((timestamp,'RWL'))    
    
for timestamp in behave['ROmL']['Time_S']:
    eve_list.append((timestamp,'ROL'))
for timestamp in behave['ROmR']['Time_S']:
    eve_list.append((timestamp,'ROR')) 
swpoint=[]

for timestamp in behave['LickR']['Time_S']:
    eve_list.append((timestamp,'lkR')) 

eve_list.sort(key=lambda x: x[0]) 
```

```python
#统计modLR的输入与其前后一位信号的时间差(确认时间差,后面可以不用计算了)
diff_pre = []
for i in range(1, len(eve_list)):
    current_id = eve_list[i][1]
    if (current_id == 'mdL' or current_id == 'mdR'):
        current_time = eve_list[i][0]
        prev_time = eve_list[i-1][0]
        diff_pre.append(current_time - prev_time)
diff_pos = []
for i in range(1, len(eve_list)):
    current_id = eve_list[i][1]
    if (current_id == 'mdL' or current_id == 'mdR'):
        current_time = eve_list[i][0]
        post_time = eve_list[i+1][0]
        diff_pos.append(post_time - current_time)
print(diff_pre)
print(diff_pos)
```

```python
#统计选择正确率

counts=[]

hit=0
count=0

mod=0

for i in eve_list:
    match i[1]:
        case 'mdL':  
            print([hit,count])
            if mod == 1:
                counts.append([hit,count])
            elif mod == -1:
                print("Error")
            hit=0
            count=0
            mod = -1
        case 'mdR':  
            print([hit,count])
            if mod == -1:
                counts.append([hit,count])
            elif mod == 1:
                print("Error")
            hit=0
            count=0
            mod = 1
        case 'RWL' | 'ROL' :
            if mod == -1:
                hit += 1
                count += 1
            elif mod == 1:
                count += 1
        case 'RWR' | 'ROR' :
            if mod == 1:
                hit += 1
                count += 1
            elif mod == -1:
                count += 1
counts.append([hit,count])
```

```python
#统计切换时间点
#统计选择正确率
SWpoints=[]

swpoint=[]

mod=0
last=0
count=0  # 正数表示进入正确边,负数表示错误边
for i in eve_list:
    match i[1]:
        case 'mdL': 
            print(swpoint)
            if mod == 1:
                SWpoints.append(swpoint)
            elif mod == -1:
                print("Error")
            swpoint=[]
            mod = -1
            last= 1
            count = 0
        case 'mdR':  
            print(swpoint)
            if mod == -1:
                SWpoints.append(swpoint)
            elif mod == 1:
                print("Error")
            swpoint=[]
            mod = 1
            last= -1
            count = 0
        case 'RWL' | 'ROL' :
            count +=1
            if last == 1 and mod == -1:
                swpoint.append(count)
            elif last == 1 and mod == 1:
                swpoint.append(-count)
            last = -1
        case 'RWR' | 'ROR' :
            count +=1
            if last == -1 and mod == 1:
                swpoint.append(count)
            elif last == -1 and mod == -1:
                swpoint.append(-count)
            last = 1
SWpoints.append(swpoint)
```
